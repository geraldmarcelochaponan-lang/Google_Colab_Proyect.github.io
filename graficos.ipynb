```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun',
         'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']

ventas_array = [150, 180, 220, 210, 250, 300, 320, 310, 350, 400, 420, 480]
gastos_array = [100, 110, 130, 120, 150, 160, 180, 170, 200, 220, 230, 250]

df = pd.DataFrame({
    'Mes': meses,
    'Ventas': ventas_array,
    'Gastos': gastos_array
})

df['Ganancia'] = df['Ventas'] - df['Gastos']

df = df.astype({'Ventas': int, 'Gastos': int, 'Ganancia': int})

print("--- TABLA DE DATOS ---")
print(df.to_string(index=False))
print("\n")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Análisis de Ventas y Ganancias Anuales', fontsize=16, fontweight='bold')

axes[0].plot(df['Mes'], df['Ventas'], marker='o', color='blue', linewidth=2, label='Ventas')
axes[0].plot(df['Mes'], df['Gastos'], marker='s', color='red', linewidth=2, linestyle='--', label='Gastos')
axes[0].set_title('Evolución Mensual (Ventas vs Gastos)', fontsize=12)
axes[0].set_ylabel('Miles de $', fontsize=10)
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.6)
axes[0].tick_params(axis='x', rotation=45)

colores = plt.cm.Greens(df['Ganancia'] / max(df['Ganancia']))
bars = axes[1].bar(df['Mes'], df['Ganancia'], color=colores, edgecolor='black')
axes[1].set_title('Ganancia por Mes', fontsize=12)
axes[1].set_ylabel('Miles de $', fontsize=10)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)
axes[1].tick_params(axis='x', rotation=45)

for bar in bars:
    yval = bar.get_height()
    axes[1].text(bar.get_x() + bar.get_width()/2, yval + 3, f'{yval}',
                 ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

# Documentación del Programa

## Objetivo
Desarrollar un panel de visualización utilizando las librerías **NumPy**, **Pandas** y **Matplotlib** en Google Colab, con la finalidad de organizar y representar datos de ventas, gastos y ganancias mediante dos tipos de gráficos. Para ello, se trabajó con arreglos de datos previamente definidos y únicamente con valores numéricos enteros.

## Herramientas Utilizadas
- **NumPy:** Se utilizó para crear y trabajar con los arreglos numéricos que contienen los datos de ventas y gastos.
- **Pandas:** Permitió organizar la información en un DataFrame, calcular las ganancias de cada mes y mantener los valores numéricos en formato entero (`int`).
- **Matplotlib:** Se empleó para crear los gráficos y presentar la información de una manera más ordenada y fácil de interpretar.

## Explicación del Código

1. **Definición de los datos:** Primero, se crearon dos arreglos llamados `ventas_array` y `gastos_array`, donde se ingresaron los valores correspondientes a los 12 meses del año. También se agregó una lista con los nombres de cada mes para identificar los resultados.

2. **Organización de la información:** Después, se utilizó Pandas para reunir los datos en un DataFrame. A partir de las ventas y los gastos, se calculó la ganancia mensual mediante una resta. Además, se convirtieron las columnas numéricas al tipo entero utilizando `.astype(int)`.

3. **Creación del gráfico de líneas:** Se elaboró el primer gráfico con `axes[0].plot()`, que permite observar cómo fueron variando las ventas y los gastos durante los 12 meses. También se agregaron colores, marcadores y una leyenda para diferenciar ambas líneas.

4. **Creación del gráfico de barras:** Para el segundo gráfico, se utilizó `axes[1].bar()` con el propósito de representar las ganancias obtenidas en cada mes. Se asignaron diferentes colores según los valores y se colocaron las cantidades encima de cada barra para facilitar su lectura.

5. **Presentación de los gráficos:** Finalmente, se utilizó `plt.subplots(1, 2)` para mostrar ambos gráficos dentro de una misma figura, uno al lado del otro. También se aplicó `plt.tight_layout()` para distribuir correctamente los elementos y evitar que se superpongan.

## Conclusión
