```markdown
# Clasificador de Reseñas de Librería basado en Palabras Clave

Este modelo clasifica las opiniones de los clientes utilizando listas extensas de palabras clave específicas del ámbito de las librerías, libreros y lectura.
```

````python
# Bloque 1: Diccionario de Palabras Clave Ampliado (Mínimo 30 palabras por categoría)
palabras_clave = {
    'Positivo': [
        'excelente', 'bueno', 'encantó', 'maravilloso', 'increíble', 'recomiendo',
        'obra', 'arte', 'joya', 'fascinante', 'rápido', 'perfecto', 'amable', 'atención',
        'maravilla', 'magnífico', 'atrapante', 'sublime', 'joyita', 'imprescindible',
        'amabilidad', 'rápida', 'impecable', 'espectacular', 'adictivo', 'precioso',
        'bello', 'gustó', 'recomendable', 'rápidez', 'fantástico', 'maravillada',
        'atento', 'cordial', 'eficaz', 'maravillosamente', 'ejemplar', 'grato'
    ],
    'Neutral': [
        'aceptable', 'normal', 'común', 'regular', 'esperado', 'cumple', 'mediocre',
        'libro', 'páginas', 'tapa', 'papel', 'precio', 'opinión', 'entrega',
        'portada', 'hojas', 'edición', 'autor', 'título', 'género', 'historia',
        'lectura', 'trama', 'narrativa', 'capítulo', 'volumen', 'ejemplar', 'envío',
        'paquete', 'comentario', 'compra', 'adquisición', 'vendedor', 'proceso',
        'adecuado', 'sencillo', 'típico', 'promedio', 'estándar', 'esperable'
    ],
    'Negativo': [
        'malo', 'aburrido', 'decepcionante', 'roto', 'dañado', 'tardó', 'caro',
        'pésimo', 'horrible', 'sucio', 'incompleto', 'lento', 'error', 'defecto',
        'decepción', 'malísimo', 'desastroso', 'demora', 'carísimo', 'arruinado',
        'aplastado', 'mojado', 'manchado', 'antiguo', 'erróneo', 'equivocado',
        'incómodo', 'pesado', 'tedioso', 'retraso', 'retrasado', 'defectuoso',
        'pobre', 'insulso', 'plagio', 'caras', 'mal', 'desagradable'
    ]
}

# Verificar que cumplimos con el requisito de más de 30 palabras clave por categoría
for cat, palabras in palabras_clave.items():
    print(f"Categoría '{cat}': {len(palabras)} palabras clave cargadas.")
````

````python
# Bloque 2: Preparación de Datos, Vectorización y Entrenamiento del Modelo
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Estructurar los datos de entrenamiento
X_train = []
y_train = []

for categoria, palabras in palabras_clave.items():
    for palabra in palabras:
        X_train.append(palabra)
        y_train.append(categoria)

# Inicializar vectorizador simple de palabras individuales (Bolsa de Palabras)
vectorizer = CountVectorizer(lowercase=True)
X_train_counts = vectorizer.fit_transform(X_train)

# Entrenar el modelo Naive Bayes Multinomial
modelo = MultinomialNB()
modelo.fit(X_train_counts, y_train)
print("¡Modelo de Naive Bayes entrenado exitosamente con las nuevas palabras clave!")
````

````python
# Bloque 3: Definición de la Función de Predicción con Detección
def clasificar_por_palabras(texto):
    # Obtener el vocabulario que conoce nuestro vectorizador
    vocabulario = set(vectorizer.get_feature_names_out())

    # Limpiar signos básicos y separar el texto por palabras individuales
    limpieza_signos = texto.lower().replace('.', '').replace(',', '').replace('!', '').replace('?', '')
    palabras_texto = set(limpieza_signos.split())

    # Detectar qué palabras del texto coinciden con las del diccionario
    palabras_detectadas = palabras_texto.intersection(vocabulario)

    # Predecir con el modelo
    texto_counts = vectorizer.transform([texto])
    prediccion = modelo.predict(texto_counts)[0]
    probabilidades = modelo.predict_proba(texto_counts)[0]

    # Obtener la probabilidad de la clase predicha
    clases = modelo.classes_
    indice_prediccion = list(clases).index(prediccion)
    probabilidad_final = probabilidades[indice_prediccion] * 100

    # Mostrar el desglose de la clasificación
    print(f"Reseña: '{texto}'")
    print(f"Palabras clave detectadas: {list(palabras_detectadas) if palabras_detectadas else 'Ninguna'}")
    print(f"Categoría predicha: {prediccion} ({probabilidad_final:.2f}% de probabilidad)")
    print("-" * 70)

    return prediccion, probabilidad_final
````

````python
# Bloque 4: Ejemplos de Validación de los Resultados
print("=== PRUEBAS DEL NUEVO MODELO DE CLASIFICACIÓN ===\n")

# Ejemplo Positivo
reseña_pos = "Esta joyita de edición es una obra de arte, me encantó la compra y su entrega fue muy rápida."
clasificar_por_palabras(reseña_pos)

# Ejemplo Neutral
reseña_neu = "El precio de este volumen es el estándar, el libro tiene un total de 400 páginas."
clasificar_por_palabras(reseña_neu)

# Ejemplo Negativo
reseña_neg = "El envío se retrasó muchísimo, el libro llegó mojado y la portada está arruinada. Qué decepción."
clasificar_por_palabras(reseña_neg)
````

## Contexto del Proyecto y Algoritmo Desarrollado

En el entorno de una **librería** (física o digital), conocer la opinión y experiencia de los lectores es fundamental para mejorar tanto la oferta de títulos como el servicio de atención y envío.

Frente a este desafío, desarrollamos un **algoritmo de clasificación de texto basado en Naive Bayes Multinomial** que analiza palabras clave estratégicas. El proceso consiste en:
1. **Definir un Diccionario Literario:** Listas específicas de palabras clave para las categorías *Positivo*, *Neutral* y *Negativo* (por ejemplo, vocablos sobre portadas, envíos, hojas, experiencia de lectura, etc.).
2. **Vectorización:** Conversión del texto en conteos numéricos de palabras individuales mediante una Bolsa de Palabras (`CountVectorizer`).
3. **Clasificación Probabilística:** El clasificador calcula estadísticamente la probabilidad de que una reseña pertenezca a cada categoría según la coincidencia y peso de las palabras detectadas.
